# Calling Model

```python
# importing the model

from langchain_community.chat_models import ChatLlamaCpp

model = ChatLlamaCpp(model_path="D:/11_Qwen_model/Qwen2.5-1.5B-Instruct-Q2_K.gguf", 
                 verbose=False,
                 temperature = 0,
                 n_ctx = 2048,
                 n_gpu_layers=-1
            )
```

# PromptTemplate

```python
# importing PromptTemplate libray
from langchain_core.prompts import PromptTemplate
```

##### PromptTemplate with one input variable

```python
# prompttemplate with one input variable

prompt_template = PromptTemplate(input_variables = ['country_name'],
                                 template= "Tell me about {country_name} country."
                                )

formatted_prompt_template = prompt_template.format(country_name = "India")

response_template = model.invoke(formatted_prompt_template)

print(response_template.content)
```

##### PromptTemplate with multi input variables

```python
# prompttemplate with multiple input variables

prompt_template_multi_input = PromptTemplate(input_variables= ['country_name', 'city_name'],
                                            template = 'Tell me about {country_name} particulary {city_name}')

formatted_prompt_template_multi_input = prompt_template_multi_input.format(country_name = 'India', city_name = 'Pune')

response_multi_input = model.invoke(formatted_prompt_template_multi_input)

print(response_multi_input.content)
```

# ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
chatprompt_template = ChatPromptTemplate([
    ("system", "You are good at {topic}. Explain them a concept about basic {subtopic}."),
    ("human", "Can you please explain me about {concept}?")
])

formatted_chatprompt_template = chatprompt_template.invoke({"topic" : "mathematics", 
                                                           "subtopic" : "algebra", 
                                                           "concept" : "multiplication"})

response_chatprompt = model.invoke(formatted_chatprompt_template)

print(response_chatprompt.content)
```

# MessagesPlaceholder

```python
# storing chat_history 

from langchain_core.messages import HumanMessage, AIMessage

chat_history = [
    HumanMessage(content="Which number is larger - 2 or 10"),
    AIMessage(content="The number 10 is larger than 2")
]
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.prompts import MessagesPlaceholder

chatprompt_template = ChatPromptTemplate([
    ("system", "You are a helpful AI assistant. Use previous conversations to answer the user's question. Perform the calculation carefully and give the final answer."),
    MessagesPlaceholder("chat_history"),
    ("human", "{question}")
    ])

formatted_chatprompt_template = chatprompt_template.invoke({"chat_history" : chat_history,
                                                            "question" : "If we multiply the larger number with 2 then what do we get? "})

response_messageplaceholder = model.invoke(formatted_chatprompt_template)

print(response_messageplaceholder.content)
```

### How do we append data into chat_history

```python
chat_history.append(HumanMessage(content="he is good person."))
chat_history.append(AIMessage(content="he is good at mathematics."))

for i in chat_history:
    print(i, '\n')
```
