# Calling Model

```python
# importing the model

from langchain_community.chat_models import ChatLlamaCpp

model = ChatLlamaCpp(model_path="D:/11_Qwen_model/Qwen2.5-1.5B-Instruct-Q2_K.gguf", 
                 verbose=False,
                 temperature = 0
                    )
```

# Structured output without input variable

```python
from pydantic import BaseModel, Field
from langchain_core.prompts import PromptTemplate

class Person(BaseModel):
    name : str = Field(description= "Name of the person")
    age  : int = Field(description= "Age of the person")
    city : str = Field(description= "Name of the city")
        
structured_output = model.with_structured_output(Person)

prompt_template = PromptTemplate(template = "He is Amey. He is 28 years old and he lives in Mumbai")

formatted_template = prompt_template.invoke({})

response_structured_output = structured_output.invoke(formatted_template)

print(response_structured_output)
```

# Structured output with input variables

```python
from pydantic import BaseModel, Field
from langchain_core.prompts import PromptTemplate

class Person(BaseModel):
    name : str = Field(description= "Name of the person")
    age  : int = Field(description= "Age of the person")
    city : str = Field(description= "Name of the city")
        
structured_output = model.with_structured_output(Person)

prompt_template = PromptTemplate(template = "He is Amey. He is 28 years old and he lives in Mumbai, {country_name}",
                                input_variables = ["country_name"])

formatted_template = prompt_template.format(country_name = "India")

response_structured_output = structured_output.invoke(formatted_template)

print(response_structured_output)
```

# Structured output - Extract multiple output

```python
from pydantic import BaseModel, Field
from langchain_core.prompts import PromptTemplate

class Person(BaseModel):
    name : str = Field(description= "Name of the person")
    age  : int = Field(description= "Age of the person")
    city : str = Field(description= "Name of the city")

class People(BaseModel):
    people: list[Person] = Field(
        description="List of all people mentioned in the text"
    )
        
structured_output = model.with_structured_output(People)

prompt_template = PromptTemplate(template = "{question}",
                                input_variables = ["question"])

formatted_template = prompt_template.format(question = "Rajesh and Ram are from Pune and Mumbai respectively. Rajesh is 25 and Ram is 29 ")

response_structured_output = structured_output.invoke(formatted_template)

print(response_structured_output)
```
