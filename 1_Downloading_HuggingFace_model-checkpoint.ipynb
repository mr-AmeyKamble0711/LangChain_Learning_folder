```python
# Downloading the Qwen model

from huggingface_hub import hf_hub_download

hf_hub_download(
    repo_id="bartowski/Qwen2.5-3B-Instruct-GGUF",
    filename="Qwen2.5-3B-Instruct-Q4_K_M.gguf",  
    local_dir="D:/11_Qwen_model"
)
```

```python
# to used the downloaded llm model using langchain framework then install the following library

pip install langchain langchain-community
```

```python
# importing the function Llama model from llama-cpp library to import the downloaded model

from langchain_community.llms import LlamaCpp

# importing file
model = LlamaCpp(model_path="D:/11_Qwen_model/Qwen2.5-3B-Instruct-Q4_K_M.gguf", 
                 verbose=False,
                 temperature = 0.15,
                 stop = ['\n']
            )

prompt1 = model.invoke("What is the capital of India?")

print(prompt1)

prompt2 = model.invoke("What is the prime minister of India?")

print(prompt2)
```
